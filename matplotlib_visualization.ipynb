# **Figures & Axes:**

### 1. Figure
* (اللوحة الكلية)هي الحاوية الكبيرة (Window / Canvas) اللي تلم كلش
* .تقدر تهز رسمة واحدة، ولا مجموعة من الرسمات المقسمة (Subplots).
* عن طريقها نتحكمو في الحجم الإجمالي تاع الصورة (figsize) والعنوان الرئيسي العالي (suptitle).


### 2. Axes
* (منطقة الرسم)هي البلاصة الحقيقية وين ينرسمو المعطيات (النقاط، الخطوط، الأعمدة...).
* كل Axes مستقل بذاته: عنده المحور X والمحور Y، العنوان الخاص بيه (set_title)، والأسماء تاع المحاور (set_xlabel).
* ملاحظة: تسمى Axes (بالجمع) لأنها تحتوي على المحورين $X$ و $Y$ معاً داخل كائن واحد.

```python

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(projection='3d')

# a 3D spiral
theta = np.linspace(0, 4 * np.pi, 200)
z = np.linspace(0, 2, 200)
r = z**0.5
x = r * np.sin(theta)
y = r * np.cos(theta)

ax.plot(x, y, z, label='spiral')
ax.legend()
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()
```

```python
np.random.seed(7)
n = 60
x = np.random.uniform(0, 10, n)
y = np.random.uniform(0, 10, n)
colors = x + y                          # used for color mapping
sizes  = np.random.uniform(20, 200, n)  # used for marker size

plt.figure(figsize=(7,6))
plt.scatter(x, y, c=colors, cmap='plasma', alpha=0.75, edgecolor='black', s=sizes)
plt.title('Random Scatter with Color/Size Mapping')
plt.xlabel('X value')
plt.ylabel('Y value')
plt.colorbar(label='X + Y') # Added color bar with label
plt.tight_layout()
plt.show()
```

```python

x = np.linspace(-4, 4, 80)
y = np.linspace(-4, 4, 80)
X, Y = np.meshgrid(x, y)
Z = np.cos(np.sqrt(X**2 + Y**2)) * np.exp(-0.1*(X**2 + Y**2))

fig = plt.figure()
ax = fig.add_subplot(projection='3d')

surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none') # Changed to plot_surface and stored for colorbar
ax.set_title('Damped Ripple Surface') # Corrected method
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.view_init(elev=35, azim=-60)
fig.colorbar(surf, shrink=0.5, aspect=10) # Attached colorbar to the surface plot
plt.show()
```

```python
labels = ['Revenue', 'Ad Spend', 'Website Visits', 'Conversion', 'Churn']
matrix = np.array([
    [1.00,  0.72,  0.65,  0.58, -0.40],
    [0.72,  1.00,  0.80,  0.30, -0.20],
    [0.65,  0.80,  1.00,  0.45, -0.15],
    [0.58,  0.30,  0.45,  1.00, -0.60],
    [-0.40,-0.20, -0.15, -0.60,  1.00],
])

plt.imshow(matrix, cmap='coolwarm', vmin=-1, vmax=1)
ticks = np.arange(len(labels))
plt.xticks(ticks, labels=labels, rotation=45, ha='right')
plt.yticks(ticks, labels=labels)

for i in range(matrix.shape[0]):
  for j in range(matrix.shape[1]):
    plt.text(j, i, f'{matrix[i, j]:.2f}', ha='center', va='center', color='black')
plt.colorbar(label='Correlation')
plt.title('Correlation Matrix')
plt.show()
```

```python
quarters  = ['Q1', 'Q2', 'Q3', 'Q4']
online    = [120, 135, 150, 170]
retail    = [90, 85, 95, 100]
wholesale = [40, 45, 38, 50]

fig, ax = plt.subplots(figsize=(7,5))

# Layer 1: starts at 0 (default)
ax.bar(quarters, online, label='Online', color='#5DA5DA')

# Layer 2: starts where layer 1 ends → bottom=online
ax.bar(quarters, retail, bottom=online, label='Retail', color='#FAA43A')

# Layer 3: starts where layers 1+2 end → bottom must be their SUM
bottom2 = np.array(online) + np.array(retail)
ax.bar(quarters, wholesale, bottom=bottom2, label='Wholesale', color='#60BD68')

ax.set_title('Revenue by Channel per Quarter')
ax.set_ylabel('Revenue ($k)')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
months      = ['Jan','Feb','Mar','Apr','May','Jun']
revenue     = [220, 245, 210, 260, 300, 280]
growth_rate = [5.2, 11.4, -14.3, 23.8, 15.4, -6.7]

fig, ax1 = plt.subplots(figsize=(8,5))

# --- Left axis: revenue as bars ---
ax1.bar(months, revenue, color='steelblue', alpha=0.7, label='Revenue')
ax1.set_xlabel('Month')
ax1.set_ylabel('Revenue ($k)', color='steelblue')
ax1.tick_params(axis='y', labelcolor='steelblue')

# --- Right axis: growth rate as a line, sharing the same x ---
ax2 = ax1.twinx()
ax2.plot(months, growth_rate, color='darkorange', marker='o', linewidth=2, label='Growth Rate')
ax2.axhline(0, color='gray', linestyle=':', linewidth=1)
ax2.set_ylabel('Growth Rate (%)', color='darkorange')
ax2.tick_params(axis='y', labelcolor='darkorange')

fig.suptitle('Monthly Revenue vs Growth Rate')
plt.tight_layout()
plt.show()
```
